```python
import pandas as pd

df_teslim = pd.read_excel('../data/raw/dogo_teslim_edilenler.xlsx')
df_iade   = pd.read_excel('../data/raw/dogo_iade.xlsx')
df_iptal  = pd.read_excel('../data/raw/dogo_iptal.xlsx')
df_aciklama = pd.read_excel('../data/raw/dogo_iade_aciklamali.xlsx')


set_teslim = set(df_teslim['Sipariş No'].dropna())
set_iade = set(df_iade['Sipariş No'].dropna())
set_iptal = set(df_iptal['Sipariş No'].dropna())

set_ana_veriler = set_teslim.union(set_iade).union(set_iptal)
set_aciklama = set(df_aciklama['Sipariş No'].dropna())

#Kesişim
aciklama_in_teslim = set_aciklama.intersection(set_teslim)
aciklama_in_iade = set_aciklama.intersection(set_iade)
aciklama_in_iptal = set_aciklama.intersection(set_iptal)

kayip_siparisler = set_aciklama.difference(set_ana_veriler)

# Kümelerden (Set) rastgele 5 adet örnek seç
def ornek_getir(veri_kumesi, adet=5):
    if len(veri_kumesi) == 0:
        return "Yok"
    return ", ".join([str(x) for x in list(veri_kumesi)[:adet]])

# Tablo 1: Sayısal Özet ve Örnekler Tablosu
ozet_data = {
    'Metrik (Kontrol Noktası)': [
        'Açıklamalı İade Dosyasındaki Toplam Sipariş',
        'Ana Verilerde (Teslim+İade+İptal) Toplam Benzersiz Sipariş',
        'Açıklamalıda Olup TESLİM Dosyasında Bulunanlar',
        'Açıklamalıda Olup İADE Dosyasında Bulunanlar',
        'Açıklamalıda Olup İPTAL Dosyasında Bulunanlar',
        'SADECE Açıklamalıda Olan Kayıp Siparişler)'
    ],
    'Adet': [
        len(set_aciklama),
        len(set_ana_veriler),
        len(aciklama_in_teslim),
        len(aciklama_in_iade),
        len(aciklama_in_iptal),
        len(kayip_siparisler)
    ],
    'Örnek Sipariş Numaraları (İlk 5 Adet)': [
        "-", 
        "-",
        ornek_getir(aciklama_in_teslim),
        ornek_getir(aciklama_in_iade),
        ornek_getir(aciklama_in_iptal),
        ornek_getir(kayip_siparisler)
    ]
}
df_ozet = pd.DataFrame(ozet_data)

# Tablo 2: Kayıp Siparişlerin Tüm Detayları
df_kayip = df_aciklama[df_aciklama['Sipariş No'].isin(kayip_siparisler)]

output_path = '../data/processed/00_Veri_Analizi_Raporu.xlsx'

with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
    df_ozet.to_excel(writer, sheet_name='Yönetici Özeti', index=False)
    df_kayip.to_excel(writer, sheet_name='Kayıp Siparişler', index=False)

print("Analiz ve Örnekleme Tamamlandı!")
print(f"Rapor Kaydedildi")
```
